## Calculos elementales

SageMath se puede utilizar como una calculadora. Las operaciones aritméticas básicas se realizan con los operadores `+`, `-`, `*`, `/`. Para la exponenciación se usa el circunflejo `^` o el doble asterisco, `**`.

```python
2+2
```

La prioridad de los operadores es la habitual:
1. Primero se agrupan los cálculos entre paréntesis, 
2. luego se evalúa las exponenciaciones (si hay varias seguidas, de derecha a izquierda),
3. luego la resta unaria, 
4. luego las multiplicaciones y divisiones (si hay varias, de izquierda a derecha),
5. luego las sumas y restas (si hay varias, de izquierda a derecha).

```python
2^2^3 # equivale a 2^(2^3), porque se hacen de derecha a izquierda
```

```python
1/2*3 # equivale a (1/2)*3, porque se hacen de izquierda a derecha
```

```python
1/2+3 # equivale a (1/2)+3, porque la división tiene prioridad sobre la suma
```

Debemos usar paréntesis para alterar la prioridad natural, pero sin abusar de ellos cuando no sean necesarios, pues dificultan la lectura.

```python
1/(2+3) # BIEN, si se quiere calcular 1/5
```

```python
((1)/(2+3)) # REGULAR: aunque también calcula 1/5, hay paréntesis innecesarios
```

Por defecto, los cálculos se realizan de forma exacta con cálculo simbólico, como se puede apreciar con `parent`.

```python
sqrt(2), parent(sqrt(2))
```

En cálculo numérico se trabaja con números de coma flotante de doble precisión de la clase `RDF` (que equivale a `double` en Python). Los cálculos con RDF son mucho más rápidos que los hechos con cálculo simbólico.

```python
RDF(sqrt(2))
```

En SageMath podemos usar la notación científica.

```python
1.24e-3, 1.24*10^(-3) # OJO: "e" significa el "exponente de potencia de 10", no es el número e
```

Existen varias constantes simbólicas predefinidas, tales como $\mathrm{e}$, $\mathrm{pi}$, el número imaginario $\mathrm{i}$ e  indeterminación $\mathrm{NaN}$.

```python
e, pi, i, NaN
```

En RDF también podemos trabajar con el épsilon de la máquina y el infinito (numérico).

```python
RDF.epsilon(), RDF(oo)
```

## Listas
Las listas son agrupaciones indexadas de valores.

```python
[66.25, 333, 1, 1234.5]
```

Podemos crear listas automáticas con la notación `[a, a+h .. b]`, que genera una lista de valores desde $a$ hasta, como máximo,  $b$ , distanciados por $h$ . Al usar `[a .. b]` se entiende que $h$ es 1.

```python
v = [1,3 .. 11] # como 3 es 1 + 2, la distancia entre puntos será de 2
v
```

También será posible crear una lista de valor de 0 hasta un entero $n-1$ escribiendo `range(n)`.

```python
list(range(10))
```

## Funciones externas

Podemos crear nuestras propias funciones.

```python
f = lambda x: sin(x)+cos(x)
```

```python
f(pi)
```

```python
g = lambda theta: RDF(sin(theta)**2+cos(theta)**2)
```

```python
g(4)
```

La funciones pueden tener varios argumentos y devolver varios valores (una lista). Por ejemplo, dados los coeficientes de la ecuación $ax^2+bx+c =0$, devolver las dos raíces:

$x = \dfrac{-b\pm\sqrt{b^2-4ac}}{2a}$.

```python
sol_poli_2 = lambda a,b,c: [ (-b+sqrt(b^2-4*a*c))/(2*a), (-b-sqrt(b^2-4*a*c))/(2*a) ]
```

```python
sol_poli_2(1, -8, 15)
```

Podemos evaluar una función en todas las componente de una lista con `map`.

```python
v = [1,2,3]
f = lambda x: x^2
list(map(f,v))
```
