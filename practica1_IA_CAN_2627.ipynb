# Gráficas de funciones
Para graficar funciones podemos usar el comando `plot`.

```python
plot(sin(x), 0, pi)
```

Este comando cuenta con múltiples opciones. Por ejemplo:
- Las opciones de `color` son `blue`, `green`, `red`, `cyan`, `magenta`, `yellow`, `black`, ...
- Las opciones para `linestyle` son: `'-'` (sólida), `'--'` (discontinua), `':'` (punteada), `'-.'` (punto y raya), `''` (sin línea, útil para mostrar solo los puntos).
- Las opciones para `marker` son: `'+'` (suma), `'x'` (aspa), `'.'` (punto pequeño), `'o'` (círculo), `'s'` (cuadrado), `'d'` (rombo), `'*'` (estrella), ...

```python
g1 = plot(sin(x), 0, 2*pi,                legend_label='Seno',   marker = '.')
g2 = plot(cos(x), 0, 2*pi, color='green', legend_label='Coseno', linestyle='--')
show(g1 + g2, title='Comparativa Seno vs Coseno', axes_labels=['x', 'y'])
```

## Eliminación de artefactos
Las gráficas se muestran el marco máximo que incluye todos los valores calculados:

```python
A = plot(lambda x: abs(1/x), -1, 1)
B = plot(lambda x: 10,       -1, 1, color='green')
show(A+B)
```

Podemos reducir el marco de visualización:

```python
(A+B).show(ymin=0, ymax=15)
```

Podemos tener "artefactos" en la imagen debido a asíntotas verticales.

```python
A = plot(lambda x: tan(x), 0, 2*pi)
show(A, ymin=-15, ymax=15)
```

Pueden ser eliminados manualmente escogiendo el dominio de definición de la función.

```python
tn = lambda x: tan(x)
eps = 1e-15
A = plot(tn,      0,       pi/2-eps) 
B = plot(tn,   pi/2+eps, 3*pi/2-eps)
C = plot(tn, 3*pi/2+eps,   2*pi)
show(A+B+C, ymin=-15, ymax=15)
```

Estos comandos nos permiten entender y comparar las gráficas de funciones.

```python
A1 = plot(lambda x: log(x,  2), 0, 10, legend_label='Log. base 2')
A2 = plot(lambda x: log(x),     0, 10, legend_label='Log. natural', color='green')
A3 = plot(lambda x: log(x, 10), 0, 10, legend_label='Log. base 10', color='red')
show(A1 + A2 + A3, title='Comparativa de logaritmos')
```

## Escala logarítmica
Podemos hacer gráficas con escalas logarítmicas. Observa que la función $y = x^4$ se representa como la recta $\log(y) = 4\cdot\log(x)$ en las variables $\log(x), \log(y)$

```python
A = plot(x^4, 1e-10, 1e-1)
show(A, title='y = x^4', axes_labels=['x', 'y'])
```

```python
B = plot(x^4, (x, 1e-10, 1e-1), scale="loglog")
show(B, title='log(y) = 4 log(x)', axes_labels=['log(x)', 'log(y)'])
```

## Líneas de puntos
A veces no tenemos la expresión de una función sino los puntos por los que debe pasar la gráfica. Por ejemplo, supongamos que tenemos puntos con abscisas $\mathbf{x} = (10, 11, \ldots, 20)$.    

```python
xi = [10 .. 20]
xi
```

Y con ordenadas $\mathbf{y} = \mathbf{x}+ \mathbf{r}$ estando $\mathbf{r}$ compuesto de valores aleatorios en $[0,1]$.

```python
yi = list(map(lambda x: x+random(), xi))
yi
```

 Ahora, debemos generar los puntos con dichas abscisas y ordenadas, mediante `zip`(cremallera).

```python
pts = list(zip(xi,yi))
pts
```

Para finalmente hacer la gráfica con `line`.

```python
line(pts)
```

## Cónicas
Podemos representar una cónica con su forma general o su expresión paramétrica. Por ejemplo, la cónica $x^2+y^2-4x+4y =0$ se escribe en su forma general como:

$(x - 2)^2 + (y + 2)^2 = 8$ 

y en su expresión paramétrica como:

$t\in[0,2\pi) \to (x,y) = \left(2 + 2\sqrt{2} \cos(t),  -2 + 2\sqrt{2} \sin(t)\right)$. 

```python
var('x y t')

f = lambda x,y: (x-2)^2 + (y+2)^2 - 8
P1 = implicit_plot(f(x,y), (x,-1,5), (y,-5,1), linestyle=':')

c = lambda t: [2+2*sqrt(2)*cos(t), -2+2*sqrt(2)*sin(t)]
P2 = parametric_plot(c(t), (t, 0, 2*pi), color='red')

show(P1+P2)
```
